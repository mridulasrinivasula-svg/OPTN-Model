```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================
# DATA
# ============================================

years = np.arange(1995, 2026)

W = np.array([
    43370, 48315, 53358, 58771, 64897, 70559, 75992, 79823,
    80565, 83205, 87651, 91265, 96130, 101542, 106950, 113782,
    121387, 126716, 133226, 140650, 146264, 147591, 148126,
    148501, 151469, 155495, 154530, 157100, 160238, 164097,
    170594
], dtype=float)

T = np.array([
    19396, 19765, 20316, 21526, 22030, 23274, 24243, 24916,
    25475, 27042, 28119, 28941, 28366, 27967, 28463, 28668,
    28553, 28059, 28956, 29540, 30974, 33610, 34770, 36530,
    39719, 39036, 41356, 42889, 46634, 48150, 49065
], dtype=float)

# ============================================
# INITIAL CONDITION
# ============================================

M0 = 420

# Baseline parameters
alpha_baseline = 0.001
gamma_baseline = 0.002


# ============================================
# MODEL
# ============================================

def simulate_M(alpha, gamma):

    M = np.zeros(len(years))

    M[0] = M0

    for t in range(len(years) - 1):

        M[t + 1] = (
            M[t]
            + alpha * W[t]
            - gamma * T[t]
        )

    return M


# ============================================
# ALPHA SENSITIVITY
# ============================================

alpha_values = [
    0.0005,
    0.00075,
    0.001,
    0.00125,
    0.0015
]

print("============================================")
print("ALPHA SENSITIVITY")
print("============================================")

plt.figure(figsize=(10, 6))

for alpha in alpha_values:

    M = simulate_M(
        alpha,
        gamma_baseline
    )

    print(
        "alpha =", alpha,
        "| M(2025) =", round(M[-1], 2),
        "days"
    )

    plt.plot(
        years,
        M,
        marker="o",
        label=f"alpha = {alpha}"
    )

plt.xlabel("Year")
plt.ylabel("Median Waiting Time (days)")
plt.title("Sensitivity of M(t) to Alpha")
plt.legend()
plt.grid(True)

plt.show()


# ============================================
# GAMMA SENSITIVITY
# ============================================

gamma_values = [
    0.001,
    0.0015,
    0.002,
    0.0025,
    0.003
]

print("\n============================================")
print("GAMMA SENSITIVITY")
print("============================================")

plt.figure(figsize=(10, 6))

for gamma in gamma_values:

    M = simulate_M(
        alpha_baseline,
        gamma
    )

    print(
        "gamma =", gamma,
        "| M(2025) =", round(M[-1], 2),
        "days"
    )

    plt.plot(
        years,
        M,
        marker="o",
        label=f"gamma = {gamma}"
    )

plt.xlabel("Year")
plt.ylabel("Median Waiting Time (days)")
plt.title("Sensitivity of M(t) to Gamma")
plt.legend()
plt.grid(True)

plt.show()


# ============================================
# BASELINE MODEL
# ============================================

M_baseline = simulate_M(
    alpha_baseline,
    gamma_baseline
)

print("\n============================================")
print("BASELINE MODEL")
print("============================================")

print("Alpha =", alpha_baseline)
print("Gamma =", gamma_baseline)
print("M(1995) =", M_baseline[0], "days")
print("M(2025) =", round(M_baseline[-1], 2), "days")
print("Minimum M =", round(np.min(M_baseline), 2), "days")
```
